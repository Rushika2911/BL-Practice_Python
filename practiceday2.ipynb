Program3: Compute Circumference and Area of Circle
Data Types: Python uses basic data types (such as float, int, bool, and string), which are all implemented as classes and objects. While Python implicitly assigns data types to variables, explicit type conversion is often necessary—for example, converting input from a string to a float using the float() function.

Constants: it is good practice to define constants in uppercase letters on top of the program in global scope

Operator Precedence (PEMDAS): When executing expressions, Python follows the order of operations: Parentheses, Exponents, Multiplication and Division (left to right), and Addition and Subtraction (left to right). Parentheses are used to override default precedence and ensure specific parts of an expression are evaluated first.

to improve readability, we use :.2f to print output upto 2 decimal places

```python
PI= 22/7
radius= float(input("enter the radius of the circle: "))
circumference= 2*PI*radius
area= PI*(radius **2)
print(f"circumfernce of the circle is {circumference:.2f}")
print(f"area of the circle is {area:.2f}")
```

Program 4: Find Generation

Operators:
- Arithmetic operators: +,-,*,/,//,%,**
- Logical operators: and, or, not
- Relational or comparison operators: ==, !=, >, <, >=, <= . they return boolean values True or False. == is used to check equality ie. checks value of variable. is operator is used to check identity ie. checks memory location of variable that if both are pointed to same object in the memory or not 
- Assigment operators: =, +=, -=, *=, /=, //=, **=, %=. they are used to assign values to variables. += is used to add and assign the value to the variable. it is equivalent to a= a+b. similarly for other operators


```python
year = int(input("Enter your birth year: "))

if 1946 < year <= 1964:
    generation = "Boomer"
elif 1965 <= year <= 1980:
    generation = "Generation X"
elif 1981 <= year <= 1996:
    generation = "Millennials"
elif 1997 <= year <= 2012:
    generation = "Generation Z"
elif 2013 <= year <= 2025:
    generation = "Generation Alpha"
else:
    generation = "Invalid birth year"

print(f"Birth year is: {year}")
print(f"Your generation is: {generation}")
```

Program 5: Time Conversion

```python
seconds = int(input("Enter Number of Seconds: "))

hours = seconds // 3600
minutes = (seconds % 3600) // 60
sec = seconds % 60

print(f"Time duration of {seconds} seconds in HH:MM:SS format is {hours:02d}:{minutes:02d}:{sec:02d}")
```
