Flip a coin, several times n compute count of Head: Flip a coin n times and compute what is the percentage of times head has come.

Notes:
Develop flip a coin as a function. This function randomly generates ‘Head’ or ’Tail’.
Use the In-built module - ‘random’ to generate Head or Tail.
Use ‘choice’ or ‘randint’ functions of this module
Develop a driver function say run_flip_a_coin(n) to flip n times and to get count of Heads
Input =>
Enter the number of times to flip the coin : 5

Output =>
Note: Everytime you execute the code the output will vary.

```
Enter the number of times to flip the coin : 5
After flipping the coin 5 times, the percentage of times head has come is: 40.00%
```
Hint =>
1. Use `random.choice`

````python
import random

def flip_coin():
    return random.choice(['Head', 'Tail'])

def run_flips_a_coin(n):
    count=0
    for _ in range(n):
        result= flip_coin()
        if result=='Head':
            count+=1
    percentage= (count/n)*100
    print(f"After flipping the coin {n} times, the percentage of times head has come is: {percentage}%")

n = int(input("Enter the number of times to flip the coin : "))
run_flips_a_coin(n)
````

PROG 3: Convert String Into Date Object

Convert string into a datetime object: On receiving the following date in string format, please convert it into Python’s DateTime object.
Date String = “22 Jun 2024”

Input => Date as Date String: 22 Jun 2024

Output =>

```
Date as Date String: 22 Jun 2024
<class 'str'>

Date as DateTime object: 2024-06-22 00:00:00
<class 'datetime.datetime'>

```
Hint =>
1. Use `datetime` module
2. Use `datetime.strptime()`


````python
from datetime import datetime

date_string = input("Date as Date String: ")


print(date_string)
print(type(date_string))

date_object = datetime.strptime(date_string, "%d %b %Y")


print("\nDate as DateTime object:", date_object)
print(type(date_object))
````

PROG 5: Recursive Function

Factorial Recursive Function: Write a factorial function in a recursive way.

Factorial takes input as n and returns factorial of n

Input => Enter the number to find the factorial : 5

Output =>
```
Enter the number to find the factorial : 5
Factorial of 5 is: 120

````python
def factorial(n):
    if n==1:
        return 1
    return n*factorial(n-1)

n= int(input("Enter the number to find the factorial: "))
print(f"Factorial of {n} is : {factorial(n)}")
````
