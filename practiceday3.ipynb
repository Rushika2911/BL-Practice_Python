Program 1: Greet function
Functions are used to optimize code by organizing repeated logic into reusable blocks, adhering to the "Don't Repeat Yourself" (DRY) principle. Defining and Calling Functions
- It is defined using the def keyword, followed by the function name (in snake_case), parentheses, and a colon.
- Proper indentation is mandatory after the function definition to denote the code block, otherwise, an indentation error will occur.
- A function must be called from outside its defined block to execute the code within it.
- To ensure data integrity, one can validate user input. For example, using the isalpha() method helps verify that a string input consists only of alphabetic characters.

```python
def greet_friend(name):
    if name.isalpha():
     print(f"Hi, {name}! How are you today?")

    else:
     print("Invalid name. Please enter a name using alphabetic characters only.")

name= input('Enter your name: ')
greet_friend(name)
```

Program 2: Car Loan Payment

```python
def calculate_car_loan_payment(P,Y,R):
    r= R/100/12
    n=Y*12
    M= (P*r*(1+r)**n)/((1+r)**n-1)
    return M



principal_amount= float(input('enter the principal amount in rupees: '))
years= float(input("enter the duration to pay off the loan in years: "))
interest_rate= float(input('enter the annual interest rate: '))

print(f"Monthly car loan payment: {calculate_car_loan_payment(principal_amount,years,interest_rate):.2f}")
```

Program 3: Compute Average
A list is a mutable data structure that can be declared using square brackets.
list can contain heterogeneous data ie data of diff datatypes


```python
def compute_average(number_list):
    count=len(number_list)
    sum_of_list= sum(number_list)
    if count==0:
        return 0
    else:
        return sum_of_list/count
number_list = [10,20,30,40,50]
print(f"original list: {number_list}")
print(f"average is: {compute_average(number_list)}")
```

Program 4: Return multiple values


```python
def min_max(number_list):
    minimum = min(number_list)
    maximum= max(number_list)
    print(f"Original list: {number_list}")
    print(f"minimum element: {minimum}")
    print(f"maximum element: {maximum}")
length= int(input("enter length of list: "))
number_list=[]
for i in range(length):
    ele= int(input("enter number: "))
    number_list.append(ele)
min_max(number_list)
```
